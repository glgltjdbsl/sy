```python
import cv2
cv2.namedWindow("My_Love", cv2.WINDOW_NORMAL)
img = cv2.imread("/home/gtp/robostar_ws/opencv/photo/latte.jpg", cv2.IMREAD_ANYCOLOR)

cv2.imshow("My_Love", img)

while True:
    if cv2.waitKey() == ord('q'):       #q
        cv2.moveWindow("My_Love", 100, 100)

    elif cv2.waitKey() == 27:       #esc
        cv2.destroyWindow("My_Love")
        break

    elif cv2.waitKey() == 13:       #enter
        cv2.moveWindow("My_Love", 50, 50)

    elif cv2.waitKey() == 9:        #tab
        cv2.resizeWindow("My_Love", 2000, 2000)
        cv2.waitKey(0)
    
```
