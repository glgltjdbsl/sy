```python
import cv2
import time

VIDEO = "/home/gtp/robostar_ws/opencv/movie/sy.mp4"            # 본인 영상이면 파일 이름만 변경

cap = cv2.VideoCapture(VIDEO)
fps = cap.get(cv2.CAP_PROP_FPS)
frame_count = cap.get(cv2.CAP_PROP_FRAME_COUNT)
delay = (1.0 / fps) * 1000
video_len = frame_count / fps

print(f"FPS = {fps}, 프레임 수 = {frame_count}, 영상 길이 = {video_len:.1f}초, delay = {delay:.2f}ms")

play_count = 1
count = 0
start = time.time()

if cap.isOpened():
    while True:
        movie, frame = cap.read()
        if movie:
            cv2.imshow("latte movie", frame)
            count += 1
            if cv2.waitKey(int(delay)) == 27:            # esc
                print("esc를 눌러 종료합니다.")
                break
        else:
            # 한 바퀴 재생 결과 출력
            play_time = time.time() - start
            print(f"[{play_count}번째 재생] 프레임 {count}장, 실제 재생 시간 {play_time:.1f}초, "
                  f"한 프레임당 지연 {(play_time - video_len) / count:.4f}초")

            # 처음부터 다시
            cap.release()
            cap = cv2.VideoCapture(VIDEO)
            play_count += 1
            count = 0
            start = time.time()
else:
    print("영상을 열 수 없습니다:", VIDEO)

cap.release()
cv2.destroyAllWindows()
```
