```python
plt.imshow(img)
```

```python
img_cvt = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(img_cvt)
```

```python
img_gray = cv2.imread("../photo/squares.jpg", cv2.IMREAD_GRAYSCALE)
plt.imshow(img_gray, cmap='gray')
```

```python
cv2.imwrite("../photo/squares_gray.jpg", img_gray)
```

```python
img_resize = cv2.resize(img_cvt, (1200,1200))
plt.imshow(img_resize)
```

```python
plt.imshow(cv2.flip(img_resize, 0))
```

```python
plt.imshow(cv2.flip(img_resize, 1))
```

```python
plt.imshow(cv2.flip(img_resize, -1))
```
