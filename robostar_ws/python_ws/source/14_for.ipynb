```python
student = "카리나"
index = 5

while index >= 1:
    print("{0} 학생, 급식이 준비되었습니다. {1} 번 남았어요." .format(student, index))
    index -= 1
    if index == 0:
        print("다시 줄서주세요.")
```

```python
student = "하니"
index = 1

while True:
    print("{0} 학생, 급식이 준비되었습니다. 호출 {1}회" .format(student, index))
```

```python
student = "장원영"
name = "Unknown"

while name != student:
    print("{0} 학생, 급식이 준비 되었습니다." .format(student))
    name = input("이름이 어떻게 되세요?")
```

```python
holiday = [3, 6]
snack = [7]

for student in range(1, 11):
    if student in holiday:
        continue
    elif student in snack:
        print("{0}번은 어딨니? 매점에서 혼 좀 나야겠구나" .format(student))
        break
    print("{0}번, 출석완료" .format(student))
```

```python
list_2d = [[1, 3], [5, 7], [9, 11], [13, 15]]

i=0
while i in range(len(list_2d)):
    j = 0
    while j < len(list_2d[i]):
        print(list_2d[i][j], end="")
        j += 1
    print()
    i += 1
```
