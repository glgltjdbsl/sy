```python
def info(name, age, *sns):
    print("이름 : {0}\t나이 : {1}\t" .format(name, age), end="")
    for s in sns:
        print(s, end="")
    print()

info("혜인", 16, "TitTok", "Facebook", "Twitter", "NaverBlog", "kakaoStory")
info("민지", 21, "Facebook", "Instagram")
```

```python
def language(lang1, lang2, lang3):
    print(lang1)
    print(lang2)
    print(lang3)

lang_list = ["python", "c++", "java"]
language(*lang_list)
```

```python
def language(*langs):
    for lang in langs:
        print(lang)


language("python", "c++", "java")
```
