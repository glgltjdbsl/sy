```python
# 다중 상속
class Chicken:
    def __init__(self):
        print("I like chicken")

class Pizza:
    def __init__(self):
        print("I like pizza")

class ChickenPizza(Chicken, Pizza):
    def __init__(self):
        #super().__init__()
        Chicken.__init__(self)
        Pizza.__init__(self)

menu = ChickenPizza()
```

```python
class Info:
    def __init__(self):
        self.name = '이름'
        self.age = '나이'
        self.gender = '성별'

    def test(self):
        print('강사 정보 클래스입니다!!')


class Teacher(Info):
    def __init__(self, name):
        super().__init__()
        self.name = name

    def display_name(self):
        super().test()
        print('강사명 : ', self.name)

    def display_age(self):
        print("강사 나이 : ", self.age)


c = Teacher("Celia")
c.display_name()
c.display_age()
```
