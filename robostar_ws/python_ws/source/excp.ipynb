```python
try:
    print("오늘 오전/오후 방문하신 손님들을 카운트 합니다.")
    num1 = int(input("오전 방문자 수 입력 : "))
    num2 = int(input("오후 방문자 수 입력 : "))
    print("{0} + {1} = {2}".format(num1, num2, int(num1+num2)))

except ValueError:
    print("정수만 입력해주세요")
```

```python
try:
    print("정수 나누기만 할 줄 아는 계산기입니다.")
    nums=[]
    nums.append(int(input("첫 번째 숫자 입력 : ")))
    nums.append(int(input("두 번째 숫자 입력 : ")))
    nums.append(int(nums[0]/nums[1]))
    print("{0} / {1} = {2} .format(nums[0], nums[1], nums[2])")

except ValueError:
    print("정수만 입력해주세요")

except ZeroDivisionError:
    print("숫자는 0으로 나눌 수는 없습니다")
```

```python
try:
    print("첫 번째와 두 번째 방문자에게 선물을 드립니다.")
    visitors = []
    visitors.append(input("첫 번째 방문자 입력 : "))
    visitors.append(input("두 번째 방문자 입력 : "))
    if visitors[0]=="Celia" or visitors[1]=="Celia":
        raise ValueError
    print("당첨자는 {0}, {1}입니다.".format(visitors[0], visitors[1]))

except ValueError:
    print("방문자 이벤트가 종료되었습니다.")
```

```python
class BlackListError(Exception):
    pass

try:
    print("첫 번째와 두 번째 방문자에게 선물을 드립니다.")
    visitors = []
    visitors.append(input("첫 번째 방문자 입력 : "))
    visitors.append(input("두 번째 방문자 입력 : "))
    if visitors[0]=="Celia" or visitors[1]=="Celia":
        raise BlackListError
    print("당첨자는 {0}, {1}입니다.".format(visitors[0], visitors[1]))
    
except ValueError:
    print("이름을 잘못 입력하셨습니다.")
except BlackListError:
    print("방문자 이벤트가 종료되었습니다.")
```

```python
class nonError(Exception):
    pass

bibim_udong = int(input("당일 판매 우동 개수 : "))
waiting_no = 1




try:
    while(True):
        print("[남은 우동 : {0}]".format(bibim_udong))
        order = int(input("우동 몇 그릇을 주문하실까요?"))

        if order > bibim_udong:
            print("남은 우동보다 주문량이 더 많습니다.")
            break
        elif bibim_udong == 0:
            raise nonError
        elif order <1:
            raise ValueError
        else:
            print("<대기번호 {0}> {1} 그릇 주문이 완료되었습니다.".format(waiting_no, order))
            waiting_no += 1
            bibim_udong -= order
            if bibim_udong == 0:
                print(bibim_udong)
                raise nonError

except ValueError:
    print("숫자를 입력해주세요")
except nonError:
    print("매진되었습니다.")
```
