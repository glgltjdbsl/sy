```python
from random import *

ls = []
count = 0

for i in range(0, 31):
    ls.append(i)

for j in range(1, 31):
    num = sample(ls, 1)[0]

    if 10 <= num <= 17:
        print("[당첨] {0}번 학생 (좋아하는 숫자 : {1})".format(j, num))
        count += 1
    else:
        print("[] {0}번 학생 (좋아하는 숫자 : {1})".format(j, num))

print("총 당첨 인원 : {0}명".format(count))
```

```python
from random import *

ls = []

for i in range(1, 21):
    ls.append(i)

shuffle(ls)

ch = sample(ls, 1)
ls.remove(ch[0])

co = sample(ls, 3)

print("--당첨자 발표--")
print("치킨 당첨자: {0}".format(ch[0]))
print("커피 당첨자: {0}".format(co))
print("--축하합니다--")
```

```python
def calculate(height, weight, gender):
    if gender == "남자":
        standard = height * height * 22
    else:
        standard = height * height * 21

    obesity = weight / standard * 100

    print("표준체중 : {:.2f}kg".format(standard))
    print("비만도 : {:.0f}%".format(obesity))


calculate(1.75, 70, "남자")
```

```python
class SoldOutError(Exception):
    pass

bibim_udong = 30
waiting_no = 1

while True:
    try:
        print("비빔 우동 : {0}".format(bibim_udong))
        order = int(input("우동 몇 그릇 주문하실까요? "))

        if order < 1:
            raise ValueError

        if order > bibim_udong:
            print("남은 우동보다 주문량이 많습니다.")
        else:
            print("대기번호 {0}번 {1}그릇 주문이 완료되었습니다.".format(waiting_no, order))
            waiting_no += 1
            bibim_udong -= order

        if bibim_udong == 0:
            raise SoldOutError

    except ValueError:
        print("1 이상의 정수만 입력해주세요.")

    except SoldOutError:
        print("비빔우동이 모두 판매되었습니다.")
        break
```
