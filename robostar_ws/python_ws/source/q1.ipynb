```python
# 문제 2. 운동장 이벤트 당첨자 찾기 (12점)
# ------------------------------------------------------------
def find_winners(numbers):
    winners = []

    for i in range(len(numbers)):
        num = numbers[i]

        if 10 <= num <= 17:
            winners.append(i + 1)

    return winners


def main():
    numbers = [10, 20, 17, 9]
    print(find_winners(numbers))


if __name__ == "__main__":
    main()
```
